## Second Largest Number

```python
# @title
lst = [7,37,4,2,65,4,3,23,39,45,23,76]

def getsecondlargest(lst):
  slar = None
  lar = lst[0]
  for x in lst[1:]:
    if x>lar:
      slar = lar
      lar = x

    elif(x!=lar):
      if x>slar or slar == None:
        slar=x
  return slar


print(getsecondlargest(lst))
```

## Reverse a List

```python
lst = [1,2,3,4,5]

def getrev(lst):
  s=0
  e=len(lst)-1

  while s<e:
    lst[s],lst[e] = lst[e],lst[s]
    s+=1
    e-=1
  return lst

print(getrev(lst))
```

## Find Pair with a given sum - TWO SUM

```python
a = [10,20,35,50]
target = 70

def two_sum(a,target):
  # Assuming the array is sorted, we will maintain two points at the start and at the end.
  s=0
  e=len(a)-1
  while s<e:
    sum1=a[s]+a[e]
    if sum1<target:
      s+=1
    elif(sum1>target):
      e-=1
    elif(sum1==target):
      return True
  return False


print(two_sum(a,target))

```

## Two Pointers Theory

```python
# ## Two Pointers Method

# The Two Pointers method is a technique used in algorithms, particularly in problems involving arrays or lists. It involves using two pointers (usually indices) that traverse the data structure to solve a problem efficiently. The pointers can move towards each other, away from each other, or in the same direction, depending on the specific problem.

# **How it works:**

# The core idea is to reduce the time complexity by avoiding nested loops or unnecessary traversals. By using two pointers, you can often perform operations or comparisons in a single pass through the data.

# **Types of Two Pointers:**

# 1.  **Pointers moving towards each other:** This is common in problems where you need to find a pair of elements that satisfy a certain condition, often in a sorted array. One pointer starts at the beginning and the other at the end, and they move inwards based on the condition.

#     *   **Example:** Finding a pair with a given sum in a sorted array.

# 2.  **Pointers moving in the same direction:** This is useful for problems involving subarrays, sliding windows, or removing duplicates. Both pointers start at the beginning, and one pointer moves faster than the other.

#     *   **Example:** Removing duplicates from a sorted array.

# **Example (Pointers moving towards each other): Finding a pair with a given sum in a sorted array**

# Let's say you have a sorted array `arr` and a target sum `target`. You want to find if there is a pair of elements in the array that adds up to the target sum.
```

```python
def remove_duplicates(arr):
  if not arr:
    return 0

  write_pointer = 1 # Pointer to the next position to write a unique element
  for read_pointer in range(1, len(arr)):
    if arr[read_pointer] != arr[read_pointer - 1]:
      arr[write_pointer] = arr[read_pointer]
      write_pointer += 1

  return write_pointer # Return the new length of the array

# Example usage
arr = [1, 1, 2, 2, 3, 4, 4, 5]
new_length = remove_duplicates(arr)
print(f"Array after removing duplicates (new length: {new_length}): {arr[:new_length]}")
```

## Two Pointers: Pointers Moving in the Same Direction



```python
# This type of Two Pointers method uses two pointers that typically start at or near the beginning of the data structure and move towards the end. The key is that the pointers might move at different rates or serve different purposes.

# **How it works:**

# One common scenario is using a "read" pointer and a "write" pointer. The read pointer iterates through the original data, while the write pointer indicates where the next element should be placed in the modified data structure. This is particularly useful for in-place operations.

# Another common scenario is the "sliding window" technique, which is a variation of this type. Here, two pointers define a "window" within the data structure. The window can expand or shrink as the pointers move, and operations are performed on the elements within the current window.

# **Patterns and Problem Types:**

# *   **Removing Duplicates:** As shown in the previous example, you can use a read pointer to iterate through the array and a write pointer to keep track of the position for unique elements. If the element at the read pointer is different from the previous element (or the element at the write pointer minus one), you copy it to the write pointer's position and increment the write pointer.

# *   **Removing a Specific Element:** Similar to removing duplicates, you can use a read pointer to traverse the array and a write pointer to place elements that are *not* the one you want to remove.

# *   **Sliding Window Problems:** These problems involve finding a subarray or substring that satisfies certain conditions. One pointer marks the start of the window, and the other expands the window. The window might shrink from the start if the condition is no longer met. Examples include finding the maximum sum subarray of a given size, finding the longest substring without repeating characters, or finding the minimum window substring.

# *   **Processing Two Arrays Simultaneously (Merge-like operations):** While not strictly "in the same direction" on a single array, you can use two pointers, one for each array, to process them simultaneously, such as merging two sorted arrays into a new array.

# **Example (Removing a Specific Element):**

# Let's say you have an array `arr` and a value `val` that you want to remove in-place.
```

```python
def remove_element(arr, val):
    write_pointer = 0  # Pointer to the next position to write a non-val element
    for read_pointer in range(len(arr)):
        if arr[read_pointer] != val:
            arr[write_pointer] = arr[read_pointer]
            write_pointer += 1
    return write_pointer # Return the new length of the array

# Example usage
arr = [0, 1, 2, 2, 3, 0, 4, 2]
val = 2
new_length = remove_element(arr, val)
print(f"Array after removing {val} (new length: {new_length}): {arr[:new_length]}")
```

## Window Sliding Technique - Maximum sum of k consecutive elements

```python
arr = [1,8,30,-5,20,7]
k=3

# we will be using window sliding technique, in which we will we first calculating the first k element sum and
# declare it as max_sum then we will be sliding one by one the window and compare the current sum with
# maximum sum

def ws1(arr,k):
  max_sum = 0
  csum=0
  for i in range(k):
    csum+=arr[i]
  max_sum=csum

  for i in range(k,len(arr)):
    csum = csum+arr[i]-arr[i-k]
    max_sum = max(csum,max_sum)
    #print(csum,max_sum)
  return max_sum


print(ws1(arr,k))

```

## Maximum Difference

```python
arr = [2, 3, 10, 6, 4, 8, 1]

# here to find the maximum difference between two elements in array where j>i, we will maintain a
# minimum value element. And we will compare this min_value with each current element in each iteration
# if the current element is less then we update the min_value with current element. Now, before that, we will
# be subtracting the current element with the min_value to get the difference and we will compare this res
# with max_diff.

def getmaxdiff(arr):
  min_val = arr[0]
  max_diff = arr[1]-arr[0]

  for i in range(1,len(arr)):
    curr_diff = arr[i]-min_val
    max_diff = max(max_diff, curr_diff)

    min_val = min(min_val, arr[i])
  return max_diff

print(getmaxdiff(arr))
```

## Trapping Rain Water

```python
arr=[3,0,1,2,5]
# To solve this question, we need to calculate how much each cell can accommodate water. The first and last cell cannot hold water so we will check between 1 to n-2. Now to find capacity of each cell, we will check the left maximum tower and right maximum tower and then take the minimum of those two, since the water can be holder till the lower tower. Then we will subtract the min_val with the current cell number to find the capacity of water for the current cell.

def getcapacity(arr):

  n=len(arr)
  res=0

  for i in range(1,n-1):
    lmax = arr[i]
    for k in range(0,i):
      if arr[k]>lmax:
        lmax=arr[k]

    rmax=arr[i]

    for m in range(i+1,n):
      rmax = max(rmax,arr[m])

    res=res+min(lmax,rmax)-arr[i]
  return res


print(getcapacity(arr))
```

## Longest Even-Odd Subarray

```python
arr = [5,10,20,6,3,8]

## This will be solved based on kadane's algorithm. Keep track of the length of even-odd and compare with the current element, if the condition breaks then start from the current element.

def getmaxeo(arr):
  curr_l=1
  res=1

  for i in range(1,len(arr)):

    if ((arr[i]%2==0 and arr[i-1]%2!=0) or (arr[i]%2!=0 and arr[i-1]%2==0)):
      curr_l+=1
      res = max(res,curr_l)
    else:
      curr_l=1
  return res

print(getmaxeo(arr))
```

## Maximum Circular Sum

```python
arr=[8,-4,3,-5,4]
arr=[3,-4,5,6,-8,7]

## To calculate maximum circular sum, we will find maximum normal subarray sum and then find maximum circular subarray sum and return the larger of both. To finc the circular sum, we will use the same kadane's algorithm, we will use the idea of minimum sum of the sub array and then subtract it with overall sum, that will give us the maximum circular sum. So , we will invert the array and then apply the same kadane's algo and then we will add this to the overall sum, since we have inverted the array, therefore adding it. And now, we compare the max normal and circular and then return the result.


def findmaxsum(arr):
  max_ending = arr[0]
  res=arr[0]

  for i in range(1,len(arr)):
    max_ending = max_ending+arr[i]
    max_ending = max(max_ending, arr[i])


    res = max(res, max_ending)

  return res

print(findmaxsum(arr))

def findoverallmaxsum(arr):
  normal_max_sum = findmaxsum(arr)

  if normal_max_sum < 0:
    return normal_max_sum

  arr_sum=0

  for i in range(len(arr)):
    arr_sum+=arr[i]
    arr[i] = -1*arr[i]

  max_circular_sum = arr_sum+findmaxsum(arr)

  return max(normal_max_sum, max_circular_sum)

print(findoverallmaxsum(arr))
```

## Window Sliding Technique - Max sum of k elements

```python
arr = [1,8,30,-5,20,7]
k=3

def findmaxk(arr,k):
  csum=0
  msum=0

  for i in range(k):
    csum+=arr[i]
  msum = csum

  for i in range(k,len(arr)):
    csum = csum+arr[i]-arr[i-k]

    msum = max(msum,csum)
  return msum

print(findmaxk(arr,k))

```

## Prefix Sum

```python
arr=[2,8,3,9,6,5,4]
 #getsum(0,2) = 13
 #getsum(2,6) = 27

 #getsum(l,r)
 # To if l=0, then means we need to find sum from 0th index till rth index. Then in this case the prefix sum will be psum[r], the rth index of prefix sum array.
 # In any other case the sum for (l,r) would be psum[r] - p[l-1]

def getpsum(arr):

  psum=[None]*len(arr)
  psum[0] = arr[0]

  for i in range(1,len(arr)):
    arr[i] = arr[i]+arr[i-1]
    psum[i] = arr[i]

  return psum

def getsum(arr,l,r):

  psum = getpsum(arr)

  if l==0:
    return psum[r]

  sum1 = psum[r]-psum[l-1]
  return sum1

print(getsum(arr,1,3))

```
